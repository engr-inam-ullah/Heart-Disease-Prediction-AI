```python
X = df.drop(columns=["target", "target_original"])
y = df["target"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
numerical_features = [
    "age",
    "trestbps",
    "chol",
    "thalach",
    "oldpeak"
]

categorical_features = [
    "sex",
    "cp",
    "fbs",
    "restecg",
    "exang",
    "slope",
    "ca",
    "thal"
]

print("Numerical features:", numerical_features)
print("Categorical features:", categorical_features)
```

```python

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

numerical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully!")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler

numerical_transformer_scaled = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

preprocessor_scaled = ColumnTransformer(
    transformers=[
        ("num", numerical_transformer_scaled, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor_scaled),
        ("classifier", LogisticRegression(
            max_iter=1000,
            random_state=42
        ))
    ]
)

logistic_model.fit(X_train, y_train)

print("Logistic Regression trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

y_pred = logistic_model.predict(X_test)
y_prob = logistic_model.predict_proba(X_test)[:, 1]

print("Logistic Regression Results")
print("-" * 30)
print("Accuracy :", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred), 4))
print("Recall   :", round(recall_score(y_test, y_pred), 4))
print("F1 Score :", round(f1_score(y_test, y_pred), 4))
print("ROC-AUC  :", round(roc_auc_score(y_test, y_prob), 4))
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

print("5-Fold Stratified Cross-Validation ready!")
```

```python
cv_results_lgbm = cross_validate(
    lgbm_model,
    X,
    y,
    cv=cv,
    scoring=[
        "accuracy",
        "precision",
        "recall",
        "f1",
        "roc_auc"
    ],
    return_train_score=False
)

print("LightGBM 5-Fold Cross-Validation")
print("-" * 40)

for metric in ["accuracy", "precision", "recall", "f1", "roc_auc"]:
    scores = cv_results_lgbm[f"test_{metric}"]
    
    print(
        f"{metric.upper():10}: "
        f"{scores.mean():.4f} ± {scores.std():.4f}"
    )
```

```python
def evaluate_model_cv(model, X, y, cv):
    results = cross_validate(
        model,
        X,
        y,
        cv=cv,
        scoring=[
            "accuracy",
            "precision",
            "recall",
            "f1",
            "roc_auc"
        ],
        return_train_score=False
    )

    return {
        "Accuracy": results["test_accuracy"].mean(),
        "Precision": results["test_precision"].mean(),
        "Recall": results["test_recall"].mean(),
        "F1": results["test_f1"].mean(),
        "ROC-AUC": results["test_roc_auc"].mean()
    }


print("CV evaluation function created successfully!")
```

```python
logistic_cv_results = evaluate_model_cv(
    logistic_model,
    X,
    y,
    cv
)

print("Logistic Regression — 5-Fold CV")
print("-" * 40)

for metric, score in logistic_cv_results.items():
    print(f"{metric:10}: {score:.4f}")
```

```python
import ipywidgets as widgets
from IPython.display import display

# Title
title = widgets.HTML(
    value="<h2>🫀 Heart Disease Prediction AI</h2>"
)

# Patient inputs
age_widget = widgets.IntSlider(
    value=55,
    min=20,
    max=100,
    description="Age:"
)

sex_widget = widgets.Dropdown(
    options=[
        ("Female", 0),
        ("Male", 1)
    ],
    value=1,
    description="Sex:"
)

cp_widget = widgets.Dropdown(
    options=[
        ("Typical Angina", 1),
        ("Atypical Angina", 2),
        ("Non-anginal Pain", 3),
        ("Asymptomatic", 4)
    ],
    value=2,
    description="Chest Pain:"
)

trestbps_widget = widgets.IntText(
    value=140,
    description="BP:"
)

chol_widget = widgets.IntText(
    value=250,
    description="Cholesterol:"
)

display(
    title,
    age_widget,
    sex_widget,
    cp_widget,
    trestbps_widget,
    chol_widget
)
```

```python
fbs_widget = widgets.Dropdown(
    options=[
        ("No", 0),
        ("Yes", 1)
    ],
    value=0,
    description="Fasting BS:"
)

restecg_widget = widgets.Dropdown(
    options=[
        ("Normal", 0),
        ("ST-T Abnormality", 1),
        ("LV Hypertrophy", 2)
    ],
    value=1,
    description="Rest ECG:"
)

thalach_widget = widgets.IntText(
    value=150,
    description="Max Heart Rate:"
)

exang_widget = widgets.Dropdown(
    options=[
        ("No", 0),
        ("Yes", 1)
    ],
    value=0,
    description="Exercise Angina:"
)

oldpeak_widget = widgets.FloatText(
    value=1.0,
    description="Oldpeak:"
)

slope_widget = widgets.Dropdown(
    options=[
        ("Upsloping", 1),
        ("Flat", 2),
        ("Downsloping", 3)
    ],
    value=2,
    description="Slope:"
)

ca_widget = widgets.Dropdown(
    options=[0, 1, 2, 3],
    value=0,
    description="Major Vessels:"
)

thal_widget = widgets.Dropdown(
    options=[
        ("Normal", 3),
        ("Fixed Defect", 6),
        ("Reversible Defect", 7)
    ],
    value=3,
    description="Thal:"
)

display(
    fbs_widget,
    restecg_widget,
    thalach_widget,
    exang_widget,
    oldpeak_widget,
    slope_widget,
    ca_widget,
    thal_widget
)
```

```python
predict_button = widgets.Button(
    description="🫀 Predict",
    button_style="success",
    tooltip="Predict Heart Disease"
)

result_output = widgets.Output()

display(predict_button, result_output)
```

```python
def make_prediction(button):

    with result_output:
        result_output.clear_output()

        # Collect values from UI
        patient = pd.DataFrame([{
            "age": age_widget.value,
            "sex": sex_widget.value,
            "cp": cp_widget.value,
            "trestbps": trestbps_widget.value,
            "chol": chol_widget.value,
            "fbs": fbs_widget.value,
            "restecg": restecg_widget.value,
            "thalach": thalach_widget.value,
            "exang": exang_widget.value,
            "oldpeak": oldpeak_widget.value,
            "slope": slope_widget.value,
            "ca": ca_widget.value,
            "thal": thal_widget.value
        }])

        # Prediction
        prediction = loaded_model.predict(patient)[0]
        probability = loaded_model.predict_proba(patient)[0]

        # Display result
        if prediction == 1:
            print("❤️ Heart Disease Detected")
            print(f"Probability: {probability[1]:.2%}")
        else:
            print("💚 No Heart Disease Detected")
            print(f"Probability: {probability[0]:.2%}")


predict_button.on_click(make_prediction)

print("Prediction system connected successfully! ✅")
```
